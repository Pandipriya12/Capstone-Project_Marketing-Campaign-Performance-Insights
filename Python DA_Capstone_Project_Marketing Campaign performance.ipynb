Exploratory Data Analysis(EDA) and Visualization:

```python
# Scatter plot
plt.figure(figsize=(10,5))
sns.scatterplot(x="Acquisition_Cost",y="ROI",data=df, color='violet')
plt.title("Acquisition_Cost vs ROI")
plt.show()

```

```python
# Bar chart
plt.figure(figsize=(10,6))
sns.barplot(x="Channel_Used",y="Conversion_Rate",data=df, hue="Campaign_Type",edgecolor="black",ci=None)
plt.xlabel("Channel_Used")
plt.ylabel("Conversion_Rate")
plt.title("Avg_ConversionRate & Channel_Used by Campaign_Type")
plt.legend(title="Campaign Type",bbox_to_anchor=(1.04,1),loc="upper left")
plt.show()
```

```python
#Box Plot
plt.figure(figsize=(10,6))
sns.boxplot(x="Campaign_Type",y="Engagement_Score",data=df, palette="Set2")
plt.title("Diff Campaign type of distribution Engagement_Score")
plt.ylabel("Engagement_Score")
plt.xlabel("Campaign_Type")
plt.show()
```

```python
#Bar chart
plt.figure(figsize=(10,6))
sns.barplot(x="Company",y="ROI",data=df, palette="Set2",edgecolor='brown',ci=None)
plt.title("Avg ROI by Company")
plt.xlabel("Company")
plt.ylabel("Avg_ROI")
plt.xticks(rotation=45)
plt.show()
```

```python
#Headmap
plt.figure(figsize=(8,6))
corr = df[['Engagement_Score',"Conversion_Rate"]].corr()
sns.heatmap(corr, annot=True,cmap="coolwarm",fmt=".2f",vmin=-1,vmax=1)
plt.title("Correlation between Engagement Score and Conversion_Rate")
plt.show()
```

Customer Segmentation:

```python
#Countplot
plt.figure(figsize=(8,6))
sns.countplot(x="Target_Audience",data=df,color="cyan",palette="Set2")
plt.xlabel("Target_Audience")
plt.ylabel("Count")
plt.title("Distribution of Target_Audience")
plt.show()
```

```python
#Which customer_segment has the highest conversion_Rate for each Language
#using Bar chart
plt.figure(figsize=(10,6))
sns.barplot(x="Customer_Segment",y="Conversion_Rate",data=df,hue="Language",estimator='max',palette="Set2",errorbar=None)
plt.xlabel("Customer_Segment")
plt.ylabel("Highest Conversion_Rate")
plt.title("Highest Conversion Rate by Customer Segment")
plt.legend(title="Language",bbox_to_anchor=(1.01,1),loc="upper left")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="Customer_Segment",y="Acquisition_Cost",data=df,hue="Channel_Used",palette="Set3")
plt.xlabel("Customer_Segment")
plt.ylabel("Acquisition Cost")
plt.title("Each Customer segment by Channel Used & Acquisition Distribution")
plt.legend(title="Channel_Used",bbox_to_anchor=(1.03,1),loc="upper left")
plt.show()
```

```python
# Average Conversion Rate by Language using bar chart
plt.figure(figsize=(8,6))
sns.barplot(x="Language",y="Conversion_Rate",data=df,errorbar=None,palette="Set1")
plt.xlabel("Language")
plt.ylabel("Avg_Conversion_Rate")
plt.title("Avg Conversion Rate by Language")
plt.show()
```

Channel Effectiveness:

```python
plt.figure(figsize=(10,6))
sns.barplot(x="Channel_Used",y="Engagement_Score",data=df,hue="Campaign_Type",ci=None)
plt.title("Different Channel & Engage Score by Campaign")
plt.legend(title="Campaign Type",bbox_to_anchor=(1.05,1),loc="upper left")
x = plt.xticks(rotation=60)
plt.show()
```

```python
#pie Chart
Total_ROI =(df.groupby('Channel_Used')['ROI'].sum())
plt.figure(figsize=(8,6))
plt.pie(Total_ROI,labels=Total_ROI.index,autopct='%1.1f%%',startangle=90,colors=sns.color_palette('Set2'))
plt.title("Distribution of total ROI diff Channel Used")
plt.show()
```

```python
#Scatter plot Clicks & impressions relationship between for each Channel
plt.figure(figsize=(8,6))
sns.scatterplot(x="Impressions",y="Clicks",data=df,hue="Channel_Used",palette="Set1")
plt.xlabel("Impressions")
plt.ylabel("Clicks")
plt.title("Clicks & impressions by Each Channel_Used")
plt.show()
```

Time_Based Analysis:

```python
#Duration using the Histogram
plt.figure(figsize=(8,6))
sns.histplot(data=df,x="Duration",bins=5,kde=True,color="magenta")
plt.xlabel("Duration")
plt.ylabel("Frequency")
plt.title("Distribution of Duration")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"],format="%d-%m-%Y")
Conversion_Trend = (df.groupby(["Date","Company"])["Conversion_Rate"].mean().reset_index())
plt.figure(figsize=(12,6))
sns.lineplot(data=Conversion_Trend,x="Date",y="Conversion_Rate",hue="Company",markers="o",linewidth=2.5)
plt.xlabel("Date")
plt.ylabel("Conversion_Rate")
plt.legend(title="Company",bbox_to_anchor=(1.05,1),loc="upper left")
plt.title("Conversion Rate Trend Over Time")
plt.show()
```

```python
plt.figure(figsize=(12,5))
Trend = (df.groupby("Date")["Engagement_Score"].mean().reset_index())
sns.lineplot(data=df,x="Date",y="Engagement_Score",color="navy",linewidth=1.5)
plt.xlabel("Date")
plt.ylabel("Engagement_Score")
plt.title("Engagement Score Trend by Date")
plt.show()
```

Geographic Analysis:

```python
#Bar Chart
plt.figure(figsize=(10,8))
Highest_Location = (df.groupby("Location")["Acquisition_Cost"].mean())
plt.bar(Highest_Location.index,Highest_Location.values,color="coral")
plt.xticks(rotation=45)
plt.xlabel("Location")
plt.ylabel("Highest_Acquisition_Cost")
plt.title("Acquisition Cost by Location")
plt.show()
```

```python
#Bar Chart
plt.figure(figsize=(14,8))
Conversion=(df.groupby(["Location","Target_Audience"])["Conversion_Rate"].mean().unstack())
sns.barplot(x="Location",y="Conversion_Rate",data=df,hue="Target_Audience",palette="Set1",ci=None)
plt.title("Conversion Rate by Location & Target Audience")
plt.legend(title="Target Audience",bbox_to_anchor=(1.05,1),loc="upper left")
plt.show()
```

```python
#Using pie chart ROI by Location
ROI_Location = (df.groupby("Location")["ROI"].mean())
plt.figure(figsize=(10,8))
plt.pie(ROI_Location,labels=ROI_Location.index,autopct='%1.1f%%',startangle=90,colors=sns.color_palette('Set3'))
plt.title("ROI by Location")
plt.show()
```
